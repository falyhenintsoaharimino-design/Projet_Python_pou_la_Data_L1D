## Présentation du dataset

## Partie B — Préparation

```python
# 1. Conversion correcte de la colonne Date (format M/J/AAAA)
df['Date'] = pd.to_datetime(df['Date'], format='%m/%d/%Y')
df['Date'].head()
```

```python
# 2. Création de Jour, Mois, Jour_Semaine
df['Jour'] = df['Date'].dt.day
df['Mois'] = df['Date'].dt.month
df['Jour_Semaine'] = df['Date'].dt.day_name()
df[['Date', 'Jour', 'Mois', 'Jour_Semaine']].head()
```

```python
# 3. Création de CA = Unit price * Quantity
df['CA'] = df['Unit price'] * df['Quantity']
df[['Unit price', 'Quantity', 'CA', 'Tax 5%', 'Total']].head()
```

```python
# 4. Comparaison de CA avec la colonne Total du dataset
df['Difference_CA_Total'] = (df['Total'] - df['CA']).round(4)
df[['CA', 'Tax 5%', 'Total', 'Difference_CA_Total']].head()
print(df[['Difference_CA_Total']].describe())
```

**Explication de la différence :** la colonne `Total` du dataset intègre la taxe de 5% (`Tax 5%`), alors que notre `CA` calculé (`Unit price * Quantity`) correspond au montant **hors taxe**. On vérifie ci-dessous que `Total ≈ CA + Tax 5%`, ce qui explique l'écart observé.

```python
verif = np.isclose(df['Total'], df['CA'] + df['Tax 5%'], atol=0.01)
print('Total == CA + Tax 5% pour', verif.sum(), 'lignes sur', len(df))
```

## Partie C — Analyse des ventes

```python
# 1. Chiffre d'affaires total (sur la base de Total, TTC)
ca_total = df['Total'].sum()
print(f"Chiffre d'affaires total : {ca_total:,.2f}")
```

```python
# 2. CA moyen par transaction
ca_moyen = df['Total'].mean()
print(f"CA moyen par transaction : {ca_moyen:.2f}")
```

```python
# 3. Quantité totale vendue
qte_totale = df['Quantity'].sum()
print('Quantité totale vendue :', qte_totale)
```

```python
# 4. Prix unitaire moyen
prix_moyen = df['Unit price'].mean()
print(f"Prix unitaire moyen : {prix_moyen:.2f}")
```

```python
# 5-6. Transaction la plus importante / la plus faible
trans_max = df.loc[df['Total'].idxmax()]
trans_min = df.loc[df['Total'].idxmin()]
print('Transaction la plus importante :')
print(trans_max[['Invoice ID', 'Branch', 'Product line', 'Total']])
print()
print('Transaction la plus faible :')
print(trans_min[['Invoice ID', 'Branch', 'Product line', 'Total']])
```

## Partie D — Analyse par agence

```python
analyse_branch = df.groupby('Branch').agg(
    nb_transactions=('Invoice ID', 'count'),
    quantite_totale=('Quantity', 'sum'),
    ca_total=('Total', 'sum'),
    ca_moyen=('Total', 'mean'),
    rating_moyen=('Rating', 'mean'),
).round(2)
analyse_branch
```

```python
agence_top_ca = analyse_branch['ca_total'].idxmax()
agence_top_rating = analyse_branch['rating_moyen'].idxmax()
print(f"Agence avec le CA le plus élevé : {agence_top_ca} ({analyse_branch.loc[agence_top_ca, 'ca_total']:.2f})")
print(f"Agence avec la meilleure note moyenne : {agence_top_rating} ({analyse_branch.loc[agence_top_rating, 'rating_moyen']:.2f})")
```

## Partie E — Analyse des produits

```python
analyse_produit = df.groupby('Product line').agg(
    ca_total=('Total', 'sum'),
    quantite_totale=('Quantity', 'sum'),
    prix_moyen=('Unit price', 'mean'),
    rating_moyen=('Rating', 'mean'),
).round(2)

classement_ca = analyse_produit.sort_values('ca_total', ascending=False)
classement_ca
```

```python
produit_top_ca = classement_ca.index[0]
produit_bottom_ca = classement_ca.index[-1]
produit_top_qte = analyse_produit['quantite_totale'].idxmax()
print(f"Produit générant le plus de CA : {produit_top_ca}")
print(f"Produit générant le moins de CA : {produit_bottom_ca}")
print(f"Produit vendu en plus grande quantité : {produit_top_qte}")
```

## Partie F — Analyse des clients

```python
analyse_genre = df.groupby('Gender').agg(
    nb_transactions=('Invoice ID', 'count'),
    quantite_moyenne=('Quantity', 'mean'),
    depense_moyenne=('Total', 'mean'),
    depense_totale=('Total', 'sum'),
    rating_moyen=('Rating', 'mean'),
).round(2)
analyse_genre
```

```python
analyse_client = df.groupby('Customer type').agg(
    nb_transactions=('Invoice ID', 'count'),
    quantite_moyenne=('Quantity', 'mean'),
    depense_moyenne=('Total', 'mean'),
    depense_totale=('Total', 'sum'),
    rating_moyen=('Rating', 'mean'),
).round(2)
analyse_client
```

## Partie G — Conclusion

```python
print(f"CA total                 : {ca_total:,.2f}")
print(f"CA moyen par transaction : {ca_moyen:.2f}")
print(f"Agence top CA            : {agence_top_ca}")
print(f"Agence top satisfaction  : {agence_top_rating}")
print(f"Produit top CA           : {produit_top_ca}")
print(f"Produit top quantité     : {produit_top_qte}")
```

**Conclusions et recommandations commerciales :**

1. **Les trois agences sont globalement équilibrées en volume**, mais présentent des écarts de CA et de satisfaction client (`Rating`) qui méritent un suivi individuel — l'agence la plus performante en CA peut servir de référence de bonnes pratiques pour les autres.
2. **Le CA calculé (`Unit price × Quantity`) est systématiquement inférieur à `Total`** d'un montant correspondant exactement à la taxe de 5% : ceci confirme la cohérence du dataset et valide notre méthode de calcul.
3. **Certaines catégories de produits (`Product line`) génèrent disproportionnellement plus de chiffre d'affaires** que d'autres : il est recommandé de renforcer les stocks et la mise en avant des catégories les plus performantes, et d'analyser les causes de sous-performance des catégories les plus faibles.
4. **La clientèle "Member" et la clientèle "Normal" ont des comportements d'achat différents** (dépense moyenne et totale) : un programme de fidélité ciblé sur les clients "Normal" à fort potentiel pourrait augmenter le taux de conversion en membres.
5. **La note de satisfaction (`Rating`) ne suit pas nécessairement le CA** : une agence ou une catégorie peut générer beaucoup de revenus tout en ayant une satisfaction client perfectible, ce qui signale un risque à moyen terme sur la fidélisation.
6. **La répartition Homme/Femme des dépenses est proche mais pas identique** : cela peut orienter des campagnes marketing différenciées par genre selon les catégories de produits les plus achetées par chaque segment.
