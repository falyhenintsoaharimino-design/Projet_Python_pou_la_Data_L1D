## Partie B — Nettoyage

```python
# Tableau récapitulatif AVANT nettoyage
avant = pd.DataFrame({
    'lignes': [df.shape[0]],
    'colonnes': [df.shape[1]],
    'doublons': [df.duplicated().sum()],
    'manquants_Age': [df['Age'].isnull().sum()],
    'manquants_Cabin': [df['Cabin'].isnull().sum()],
    'manquants_Embarked': [df['Embarked'].isnull().sum()],
})
avant.index = ['avant_nettoyage']
avant
```

```python
# Recherche puis suppression des doublons (hors identifiant unique PassengerId)
nb_doublons = df.drop(columns=['PassengerId']).duplicated().sum()
print('Nombre de doublons détectés (hors PassengerId) :', nb_doublons)
df = df.drop_duplicates(subset=[c for c in df.columns if c != 'PassengerId'])
print('Dimensions après suppression des doublons :', df.shape)
```

```python
# Traitement de l'Age : valeurs manquantes -> médiane par groupe (Sex, Pclass), plus robuste que la moyenne globale
df['Age'] = df.groupby(['Sex', 'Pclass'])['Age'].transform(lambda s: s.fillna(s.median()))
# S'il reste des NaN (groupe entièrement vide), on complète avec la médiane globale
df['Age'] = df['Age'].fillna(df['Age'].median())
print('Valeurs manquantes restantes pour Age :', df['Age'].isnull().sum())
```

```python
# Traitement de Cabin : trop de valeurs manquantes (>75%) pour être imputées de façon fiable.
# On ne supprime pas la colonne : on la transforme en indicateur binaire "a une cabine connue" (information utile),
# et on garde uniquement la première lettre (pont/deck) quand elle est connue.
df['Deck'] = df['Cabin'].str[0]
df['Deck'] = df['Deck'].fillna('Inconnu')
df['Cabin_connue'] = np.where(df['Cabin'].notnull(), 1, 0)
print(df['Deck'].value_counts())
```

```python
# Traitement de Embarked : seulement 2 valeurs manquantes -> imputation par le mode (valeur la plus fréquente)
mode_embarked = df['Embarked'].mode()[0]
df['Embarked'] = df['Embarked'].fillna(mode_embarked)
print('Valeur utilisée pour compléter Embarked :', mode_embarked)
print('Valeurs manquantes restantes pour Embarked :', df['Embarked'].isnull().sum())
```

```python
# Fare : pas de valeur manquante dans train.csv, on vérifie et on sécurise le traitement au cas où
print('Valeurs manquantes pour Fare :', df['Fare'].isnull().sum())
df['Fare'] = df['Fare'].fillna(df.groupby('Pclass')['Fare'].transform('median'))
```

```python
# Tableau récapitulatif APRES nettoyage
apres = pd.DataFrame({
    'lignes': [df.shape[0]],
    'colonnes': [df.shape[1]],
    'doublons': [df.drop(columns=['PassengerId']).duplicated().sum()],
    'manquants_Age': [df['Age'].isnull().sum()],
    'manquants_Cabin(brute)': [df['Cabin'].isnull().sum()],
    'manquants_Embarked': [df['Embarked'].isnull().sum()],
})
apres.index = ['apres_nettoyage']
recap = pd.concat([avant, apres])
recap
```

**Justification des choix :**
- `Age` : imputation par la médiane du sous-groupe (Sexe × Classe), plus représentative qu'une médiane globale car l'âge varie selon ces deux critères.
- `Cabin` : trop incomplète (>75% de valeurs manquantes) pour une imputation fiable ; conservée sous forme d'indicateur `Cabin_connue` et de `Deck` (pont), qui portent de l'information sans inventer de données.
- `Embarked` : seulement 2 valeurs manquantes → imputation par le mode (port le plus fréquent), impact négligeable.
- `Fare` : aucune valeur manquante ici, mais un traitement par médiane par classe est prévu par sécurité (utile notamment sur `test.csv`).

## Partie C — Création de variables

```python
# FamilySize
df['FamilySize'] = df['SibSp'] + df['Parch'] + 1

# IsAlone
df['IsAlone'] = np.where(df['FamilySize'] == 1, 1, 0)

# AgeGroup avec np.select
conditions = [
    df['Age'] < 13,
    (df['Age'] >= 13) & (df['Age'] <= 17),
    (df['Age'] >= 18) & (df['Age'] <= 59),
    df['Age'] >= 60,
]
choices = ['Enfant', 'Adolescent', 'Adulte', 'Senior']
df['AgeGroup'] = np.select(conditions, choices, default='Inconnu')

df[['Age', 'AgeGroup', 'SibSp', 'Parch', 'FamilySize', 'IsAlone']].head(10)
```

## Partie D — Analyse

```python
# 1. Taux global de survie
taux_global = df['Survived'].mean() * 100
print(f'Taux global de survie : {taux_global:.2f}%')
```

```python
# 2. Taux de survie par sexe
taux_sexe = df.groupby('Sex')['Survived'].mean() * 100
taux_sexe.round(2)
```

```python
# 3. Taux de survie par classe
taux_classe = df.groupby('Pclass')['Survived'].mean() * 100
taux_classe.round(2)
```

```python
# 4. Taux de survie selon le sexe ET la classe
taux_sexe_classe = df.groupby(['Sex', 'Pclass'])['Survived'].mean().unstack() * 100
taux_sexe_classe.round(2)
```

```python
# 5-6. Âge moyen des survivants vs non-survivants
age_moyen = df.groupby('Survived')['Age'].mean()
print(f"Âge moyen des non-survivants : {age_moyen[0]:.2f} ans")
print(f"Âge moyen des survivants     : {age_moyen[1]:.2f} ans")
```

```python
# 7. Taux de survie par groupe d'âge
taux_agegroup = df.groupby('AgeGroup')['Survived'].mean().sort_values(ascending=False) * 100
taux_agegroup.round(2)
```

```python
# 8-9. Taux de survie : seul vs en famille
taux_isalone = df.groupby('IsAlone')['Survived'].mean() * 100
taux_isalone.index = ['En famille (0)', 'Seul (1)']
taux_isalone.round(2)
```

```python
# 10. Prix moyen du billet par classe
fare_moyen_classe = df.groupby('Pclass')['Fare'].mean().round(2)
fare_moyen_classe
```

## Partie E — Conclusion

```python
print(f"Taux global de survie          : {taux_global:.1f}%")
print(f"Survie femmes / hommes          : {taux_sexe['female']:.1f}% / {taux_sexe['male']:.1f}%")
print(f"Survie 1re / 2e / 3e classe      : {taux_classe[1]:.1f}% / {taux_classe[2]:.1f}% / {taux_classe[3]:.1f}%")
print(f"Survie femmes 1re classe         : {taux_sexe_classe.loc['female', 1]:.1f}%")
print(f"Survie hommes 3e classe          : {taux_sexe_classe.loc['male', 3]:.1f}%")
print(f"Survie voyageant seul / famille  : {taux_isalone['Seul (1)']:.1f}% / {taux_isalone['En famille (0)']:.1f}%")
```

**Conclusions (interprétation des résultats ci-dessus) :**

1. **Le sexe est le facteur de survie le plus déterminant** : les femmes ont un taux de survie très largement supérieur à celui des hommes (règle « les femmes et les enfants d'abord » clairement respectée dans l'évacuation).
2. **La classe sociale a fortement influencé la survie** : les passagers de 1ʳᵉ classe ont survécu dans une bien plus grande proportion que ceux de 3ᵉ classe, probablement en raison de la proximité des cabines avec les canots de sauvetage et d'un accès prioritaire au pont.
3. **L'effet cumulé sexe + classe est majeur** : les femmes de 1ʳᵉ classe affichent le taux de survie le plus élevé du dataset, tandis que les hommes de 3ᵉ classe affichent le taux le plus faible — ces deux variables se renforcent mutuellement.
4. **Les enfants ont été globalement favorisés** : le taux de survie par groupe d'âge montre un avantage pour les plus jeunes passagers par rapport aux adultes.
5. **La composition familiale joue un rôle intermédiaire** : voyager en petite famille (ni seul, ni en très grand groupe) est associé à un meilleur taux de survie que voyager complètement seul, probablement lié à l'entraide et à la prise en charge mutuelle lors de l'évacuation.
6. **Le prix du billet reflète la hiérarchie des classes** et corrèle avec la survie : plus le tarif moyen payé est élevé, plus le taux de survie de la classe correspondante est important — le tarif est ici un indicateur indirect du statut social et de la localisation de la cabine.

**Réponse à la question posée :** les passagers ayant eu le plus de chances de survie sont les **femmes voyageant en 1ʳᵉ ou 2ᵉ classe, en famille restreinte**. À l'inverse, les **hommes adultes de 3ᵉ classe voyageant seuls** constituent le profil ayant eu le moins de chances de survie.
